```python
from pyspark.sql import SparkSession
import time

# 1. Initialize Active Distributed Spark Session
spark = SparkSession.builder \
    .appName("Interactive_Browser_Spark_Job") \
    .config("spark.sql.shuffle.partitions", "4") \
    .getOrCreate()

print("🚀 Spark Session Started successfully inside the browser!")

# CHANGE THIS LINE: Remove "work/data/" since the notebook is already inside the data folder
transactions_df = spark.read.option("multiLine", "true").json("transactions.json")
transactions_df.show(5)

# 3. Keep Spark engine alive for 10 minutes so you can view port 4040
print("⏳ Spark UI is now LIVE! Open http://localhost:4040 in a new browser tab to check it.")
time.sleep(600) 

# Stop the session when the timer finishes
spark.stop()
print("Spark Session Closed.")
```
